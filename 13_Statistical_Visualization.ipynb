Data

## 1.Histogram

A Histogram shows the distribution of numerical data

It divides values into intervals called bins and shows how many observations fall into each bin

Example:

Age

20  21  22  23  24  ...  60

A histogram can show whether the data is:

Normally distributed
Right-skewed
Left-skewed
Concentrated around a particular range
Containing possible unusual values

### When to use

Use a histogram when you want to understand:

Distribution of numerical data
Frequency
Skewness
Possible outliers
Concentration of values

### Advantages

Easy to understand
Shows distribution clearly
Helps identify skewness
Useful during EDA
Helps identify appropriate statistical methods

### Limitations:

Only suitable for numerical data.
Results depend on bin size.
It can hide details when the dataset is small.
Different bin sizes can produce different appearances.

```python
import matplotlib.pyplot as plt                                #PythonExample

age = [20, 22, 25, 27, 28, 30, 31, 32, 35, 36, 40, 42, 45]

plt.hist(age, bins=5)

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")

plt.show()
```

```python
Interpretation:

If most observations are between 25 and 35:

Most people in the dataset are between 25 and 35 years old.

If the distribution has a long tail toward higher values:

The data may be right-skewed.
```

## 2.Bar Chart

A Bar Chart compares values across categorical groups

When to use:

Use a bar chart for:

Category comparisons
Product sales
Number of customers by region
Department performance
Top-N comparisons

### Advantages

Easy to compare categories
Easy to understand
Works well with categorical data
Can show ranking clearly

### Limitations

Not ideal for continuous distributions
Too many categories make the chart crowded
Doesn't show distribution within each category


```python
import matplotlib.pyplot as plt                   #PythonExample

products = ["Laptop", "Mobile", "Tablet"]
sales = [120, 200, 80]

plt.bar(products, sales)

plt.xlabel("Product")
plt.ylabel("Sales")
plt.title("Product Sales")

plt.show()
```

```python
### Interpretation

If Mobile has the tallest bar:

Mobile has the highest sales among the three products
```

## 3.Line Chart

A Line Chart shows how a value changes across an ordered sequence, especially time

When to use

Use a line chart for:

Time-series data
Monthly sales
Stock prices
Website traffic
Temperature
Revenue trends

### Advantages

Excellent for trends
Shows increases and decreases clearly
Good for time-series data
Multiple lines can compare trends

### Limitations

Not ideal for unrelated categories
Too many lines can become confusing
Connecting unrelated observations can be misleading

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]              #PythonExample
sales = [100, 120, 150, 140, 180]

plt.plot(months, sales, marker="o")

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales")

plt.show()
```

```python
Interpretation:

If sales increase from January to May:

Sales show an overall increasing trend.
```

## 4.Scatter Plot

A Scatter Plot shows the relationship between two numerical variables

When to use

Use a scatter plot to:

Study relationships between numerical variables.
Look for correlation.
Detect clusters.
Identify possible outliers.
Check whether a relationship appears linear.

### Advantages:

Shows individual observations.
Helps identify correlation.
Shows outliers.
Shows clusters.
Useful before regression.

### Limitations:

Only two dimensions are shown easily
Large datasets can create overlapping points
Correlation does not imply causation

```python
hours = [1, 2, 3, 4, 5, 6]                             #PythonExample
scores = [45, 50, 58, 65, 72, 80]

plt.scatter(hours, scores)

plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Study Hours vs Exam Score")

plt.show()
```

```python
Interpretation:

If points generally move upward:

There appears to be a positive relationship between study hours and exam score
```

## 5.Pie Chart

A Pie Chart shows how categories contribute to a whole

When to use

Use a pie chart when:

Showing part-to-whole relationships
There are only a few categories
Categories add up to a meaningful total

### Advantages

Simple for a small number of categories
Shows proportions visually
Easy for general audiences

### Limitations

Difficult to compare similar-sized slices
Too many categories make it confusing
Not good for time-series data
Bar charts are usually better when precise comparison is important

```python
products = ["Laptop", "Mobile", "Tablet"]             #PythonExample
sales = [40, 35, 25]

plt.pie(
    sales,
    labels=products,
    autopct="%1.1f%%"
)

plt.title("Sales Distribution")

plt.show()
```

```python
Interpretation:

Laptop contributes 40% of the total sales
```

## 6.Density Plot

A Density Plot shows the estimated probability distribution of numerical data

It is similar to a smooth histogram

Instead of bars, it uses a smooth curve

When to use

Use density plots to:

Understand data distribution.
Compare distributions.
Identify peaks.
Examine skewness.
Compare groups.

### Advantages

Smooth representation
Easy to compare distributions
Doesn't depend on histogram bins
Shows peaks and shape clearly

### Limitations

The result depends on the smoothing/bandwidth
Can be confusing for beginners
Not ideal for very small datasets
Can hide individual observations

```python
import seaborn as sns                             #PythonExample
import matplotlib.pyplot as plt

salary = [25000, 28000, 30000, 32000, 35000,
          36000, 38000, 40000, 45000]

sns.kdeplot(salary, fill=True)

plt.xlabel("Salary")
plt.title("Salary Density Plot")

plt.show()
```

```python
Interpretation:

A high peak around 35,000 means:

A large concentration of salary values is around ₹35,000
```

## 7.Box Plot

A Box Plot summarizes the distribution of numerical data using:

Minimum/non-outlier lower value

.Q1
.Median
.Q3
.Maximum/non-outlier upper value
.Possible outliers

The IQR (Interquartile Range) is:

$$ IQR = Q3-Q1 $$

A common outlier rule is:

$$ Lower\ Bound = Q1 - 1.5(IQR) $$

$$ Upper Bound=Q3+1.5(IQR) $$

When to use

Use box plots to:

Detect outliers.
Compare distributions.
Understand spread.
Compare groups.

### Advantages

Excellent for detecting possible outliers
Compact summary
Easy to compare groups
Shows median and spread

### Limitations

Doesn't show every data point
Can hide multimodal distributions
Beginners may need explanation of quartiles

```python
salary = [25000, 27000, 28000, 30000, 32000,           #PythonExample
          35000, 36000, 38000, 100000]

plt.boxplot(salary)

plt.ylabel("Salary")
plt.title("Salary Box Plot")

plt.show()
```

```python
Interpretation:

If 100000 appears separately from the main box:

₹100,000 may be a potential outlier and should be investigated

Important: An outlier is not automatically an error
```

## 8. Violin Plot

A Violin Plot combines ideas from:

Box Plot
Density Plot

It shows the distribution and density of numerical data for one or more groups.

When to use

Use violin plots when:

Comparing distributions between groups
You want more distribution detail than a box plot
You want to see where observations are concentrated

### Advantages

Shows distribution shape
Shows density
Useful for comparing groups
Can reveal multiple peaks

### Limitations

More difficult to understand than a box plot
Density estimation can be sensitive to smoothing
Can be misleading with very small samples
Doesn't show exact observations by itself

```python
import pandas as pd
import seaborn as sns                          #PythomImplementation
import matplotlib.pyplot as plt

data = {
    "Department": [
        "IT", "IT", "IT", "IT",
        "HR", "HR", "HR", "HR"
    ],
    "Salary": [
        40000, 45000, 50000, 55000,
        30000, 32000, 35000, 38000
    ]
}

df = pd.DataFrame(data)

sns.violinplot(
    x="Department",
    y="Salary",
    data=df
)

plt.title("Salary Distribution by Department")

plt.show()
```

```python
Interpretation:

If the IT violin is centered higher:

IT employees have generally higher salaries than HR employees in this illustrative dataset

The width of the violin indicates where observations are more densely concentrated
```

### 9.Pair Plot

A Pair Plot displays relationships between multiple numerical variables at the same time

For example:

Age
Salary
Experience
Performance

It creates scatter plots for pairs of variables and distributions along the diagonal

When to use

Use a pair plot during EDA when you want to quickly explore:

Relationships between multiple numerical variables.

Possible correlations.

Clusters

Outliers

Distributions

### Advantages

Very useful for initial EDA

Shows many relationships at once

Helps identify patterns quickly

Can be colored by a categorical variable

### Limitations

Becomes crowded with many features

Can be computationally expensive for large datasets

Mostly useful for numerical variables

Not ideal as a final presentation visualization


```python
import seaborn as sns        #PythonExample
import matplotlib.pyplot as plt

df = sns.load_dataset("iris")

sns.pairplot(
    df,
    hue="species"
)

plt.show()
```

```python
Interpretation:

You might observe:

Some flower measurements show strong relationships, while different species form distinct clusters.

This can also provide intuition about which features might be useful for classification.
```

## 10.Heatmap

A Heatmap represents values using different levels of visual intensity

One of the most common uses in Data Science is visualizing a correlation matrix

When to use

Use a heatmap for:

Correlation matrices.

Missing-value patterns.

Confusion matrices.

Large tables of numerical values.

Feature relationships.

### Advantages

Quickly reveals patterns

Excellent for correlation matrices

Easy to identify high/low values

Can display many values compactly

### Limitations

Too many variables can make it difficult to read

Color interpretation can be misleading

Correlation heatmaps only show pairwise relationships represented by the chosen correlation measure

Correlation does not imply causation

```python
import seaborn as sns                  #PythonExample
import matplotlib.pyplot as plt

df = sns.load_dataset("iris")

correlation = df.select_dtypes(
    include="number"
).corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
Interpretation:

If:

Petal Length ↔ Petal Width = 0.96

then:

Petal length and petal width have a very strong positive linear relationship in this dataset.
```
