# Notebook 10 – Hypothesis Testing

### What is Hypothesis Testing?

Hypothesis testing is a statistical method used to test a claim about a population using sample data

Example: A college claims that the average student mark is 70. We collect marks from a sample of students to check whether the claim is reasonable

```python
marks = [65, 72, 75, 68, 80, 74, 71, 77]           #PythonImplementation

average = sum(marks) / len(marks)

print("Average Marks:", average)
```

### Null Hypothesis ($ H_0 $)

The null hypothesis is the starting assumption. It usually states that there is no change, no difference, or no effect.

Formula:

$$
H_0: \mu = \mu_0
$$

Numerical example: 

A company claims that its employees' average training score is 70. We test this claim using a sample

Business example:

A company claims its average delivery time is 3 days. The null hypothesis is H
0
	​

:μ=3 days.

AI/ML example:

A data scientist tests whether a new model's accuracy differs from the existing model's accuracy. The null hypothesis states that there is no difference.

```python
population_mean = 70                #PythonImplementation
sample_mean = 72.75

print("Claimed Mean:", population_mean)
print("Sample Mean:", sample_mean)
```

### Alternative Hypothesis ($ H_1 $) 0r ($ H_a $)

The alternative hypothesis represents the effect, difference, or change we want to investigate

Formula:

$$
H_1: \mu \neq 70
$$

Business example:

A company investigates whether its new advertising campaign changes average sales.

AI/ML example:

Test whether a new machine-learning model performs differently from the old model.

```python
claimed_mean = 70                    #pythonimplementation
sample_mean = 72.75

if sample_mean != claimed_mean:
    print("The sample mean differs from the claimed mean.")
```

### p-value

The p-value measures how surprising the observed result, or a more extreme result, would be if the null hypothesis were true.

Decision rule:

$$ p < 0.05 \implies \text{Reject } H_0 $$

Business example:

A company tests whether a new website design increases sales. A p-value of 0.02 provides evidence against the null hypothesis of no increase, assuming an appropriate test was used.

AI/ML example: 

A statistical comparison of two models gives a p-value of 0.01. This provides evidence of a difference under the test assumptions.

Remember: A p-value of 0.03 does not mean there is a 3% probability that the null hypothesis is true

```python
p_value = 0.03            #pythonimplementation
alpha = 0.05

print("P-value:", p_value)

if p_value < alpha:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

### Confidence Interval

A confidence interval estimates a range of plausible values for a population parameter.

For a population mean with unknown population standard deviation:

Confidence Interval Formula:

$$
\bar{x} \pm t \cdot \frac{s}{\sqrt{n}}
$$

Numerical example: 

Suppose a sample produces an estimated average salary of ₹30,000 and a 95% confidence interval of ₹28,000 to ₹32,000

Business example:

Estimate the average amount customers spend per order, along with a 95% confidence interval

AI/ML example: 

Estimate the mean model accuracy across repeated test samples or suitable resampling runs, with a confidence interval

```python
import numpy as np                         #PythonExample
from scipy import stats

salaries = np.array([28000, 29000, 30000, 31000, 32000])

mean = np.mean(salaries)

confidence_interval = stats.t.interval(
    confidence=0.95,
    df=len(salaries) - 1,
    loc=mean,
    scale=stats.sem(salaries)
)

print("Mean Salary:", mean)
print("95% Confidence Interval:", confidence_interval)
```

### Significance Level (α)

The significance level is the threshold chosen before testing to decide when evidence against the null hypothesis is strong enough to reject it

Common significance levels are 0.10, 0.05, and 0.01.

Numerical example: Let α=0.05 and the p-value be 0.02

Business example:

A company chooses a 5% significance level before testing a new marketing strategy

AI/ML example: 

A researcher sets α=0.01 before comparing model performance, using a stricter threshold for rejecting the null hypothesis

```python
alpha = 0.05                           #PythonExample
p_value = 0.02

if p_value < alpha:
    print("Statistically significant")
else:
    print("Not statistically significant")
```

### Type I Error

A Type I error occurs when we reject a null hypothesis that is actually true

Example:

A test concludes that a new advertising campaign improves sales when, in reality, it does not

P(Type I Error) = α

If α=0.05, the test's Type I error rate is controlled at 5% when the null hypothesis is true and the test assumptions hold.

Business example:

A company launches an expensive product because its experiment incorrectly indicates an improvement

AI/ML example:

A researcher concludes that a new model is better when there is no genuine performance improvement

### Type II Error

A Type II error occurs when we fail to reject a null hypothesis that is actually false.

Example:

A new advertising campaign genuinely improves sales, but the experiment fails to detect the improvement.

P(Type II Error)=β

Statistical power is:

Power = 1−β

Business example:

A company abandons a genuinely effective marketing strategy because its experiment does not detect the improvement

AI/ML example:

A new model performs better, but a small test dataset provides insufficient evidence to demonstrate the improvement

### One-tailed Test

A one-tailed test checks for an effect in one specified direction.

Example: Test whether the average mark is greater than 70.

H
0
	​

:μ=70
H
1
	​

:μ>70

Business example:

Test whether a new sales strategy increases average monthly sales.

AI/ML example:

Test whether a new model improves accuracy compared with a baseline, using a suitable statistical test.

```python
from scipy import stats                        #PythonExample

marks = [72, 75, 78, 74, 80, 77, 76, 79]

result = stats.ttest_1samp(
    marks,
    popmean=70,
    alternative="greater"
)

print("T-statistic:", result.statistic)
print("P-value:", result.pvalue)

if result.pvalue < 0.05:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

### Two-tailed Test

A two-tailed test checks for a difference in either direction

Example: Test whether the average mark differs from 70

H
0
	​

:μ=70
H
1
	​

:μ

=70

Business example:

Test whether a new pricing strategy changes average sales, whether sales increase or decrease

AI/ML example: 

Test whether a new model's accuracy differs from the baseline in either direction

```python
from scipy import stats                                 #PythonExample

marks = [72, 75, 78, 74, 80, 77, 76, 79]

result = stats.ttest_1samp(
    marks,
    popmean=70,
    alternative="two-sided"
)

print("T-statistic:", result.statistic)
print("P-value:", result.pvalue)

if result.pvalue < 0.05:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```
