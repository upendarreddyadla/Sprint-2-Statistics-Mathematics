# Notebook 4 – Data Distribution

Data distribution tells us how data values are spread or arranged

### Normal Distribution


```python
A Normal Distribution is a probability distribution where most values are concentrated around the mean, and fewer values occur as we move away from the mean

It has a bell-shaped curve

Simple example:

Suppose the heights of students are:

150, 155, 158, 160, 161, 162, 165, 168, 170

Most values are around the middle, while very low and very high values are less common.

Important properties:
    
1.Bell-shaped
2.Symmetric
3.Mean = Median = Mode
4.Most observations are near the mean
5.Total area under the curve = 1

Business example:

Employee salaries in a company may approximately follow a normal distribution within a particular job level

AI/ML example:

Many statistical ML techniques assume that some variables or errors are approximately normally distributed
```

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(50, 10, 1000)

plt.hist(data, bins=30)
plt.title("Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

```python
What is happening:

data = np.random.normal(50, 10, 1000)

means:

50 → Mean
10 → Standard deviation
1000 → Generate 1000 random values

The graph will be bell-shaped, with most values around 50.

Important:

Because np.random.normal() generates random data, your graph will be slightly different every time you run the cell
So the main output is a bell-shaped histogram
```

### Uniform Distribution


In a Uniform Distribution, every value within a specified range has an equal or approximately equal probability.

Simple example

Rolling a fair six-sided die:

1  2  3  4  5  6

Each number has the same probability:



$$
P(X=x)=\frac{1}{6}
$$

```python
Business example:

If a system randomly selects a number between 1 and 100, each number can have equal probability

AI/ML example:

Uniform random numbers are commonly used when initializing or generating random values in algorithms
```

```python
import numpy as np                                #PythonImplementation
import matplotlib.pyplot as plt

data = np.random.uniform(1, 10, 1000)

plt.hist(data, bins=20)
plt.title("Uniform Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

### Gaussian Distribution


Gaussian Distribution is another name for the Normal Distribution.

So:

Gaussian Distribution = Normal Distribution

It is called Gaussian after the mathematician Carl Friedrich Gauss.

### Formula

$$
f(x)=\frac{1}{\sigma\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
$$

Where:

\(\mu\) = mean
\(\sigma\) = standard deviation
\(x\) = observation

```python
Business example:

Measurement errors can often be approximately Gaussian

AI/ML example:

Gaussian assumptions are used in techniques such as Gaussian Naive Bayes
```

```python
import numpy as np

data = np.random.normal(
    loc=50,
    scale=10,
    size=1000
)

print(data[:10])
```

```python
loc = mean
scale = standard deviation
size = number of values
```

### Left Skew

```python
A Left-Skewed Distribution has a long tail toward the left side

It is also called Negative Skewness
        █████
      ███████
    █████████
  ███████████
████████████
<------------
  Long tail

Relationship

Generally:

  Mean < Median < Mode

Simple example:

Exam scores where most students score very high:

40, 70, 80, 85, 88, 90, 92, 95, 98

Most students have high scores, while a few students have very low scores

Business example:

Customer satisfaction scores may be left-skewed if most customers give high ratings

AI/ML example:

Skewed features can sometimes affect statistical models and may require transformation
```

```python
import numpy as np                      #PythonImplementation
import matplotlib.pyplot as plt

data = np.random.beta(5, 2, 1000)

plt.hist(data, bins=30)
plt.title("Left Skewed Distribution")
plt.show()
```

### Right Skew


```python
A Right-Skewed Distribution has a long tail toward the right side.

It is also called Positive Skewness

████████
██████
████
███
██
  ------------>
     Long tail

Relationship

Generally:

Mode < Median < Mean

Simple example:-

Income data:

20k, 22k, 25k, 27k, 30k, 35k, 40k, 100k

Most people have moderate income, but a few people have extremely high income

 Business example

Income, house prices, and transaction amounts are often right-skewed.

AI/ML example:

A feature such as AnnualIncome may be highly right-skewed.

A transformation such as:

np.log1p(data)

can sometimes reduce strong right skew
```

```python
import numpy as np                                  #PythonImplementation
import matplotlib.pyplot as plt

data = np.random.exponential(2, 1000)

plt.hist(data, bins=30)
plt.title("Right Skewed Distribution")
plt.show()
```

### Symmetric Distribution


```python
A distribution is symmetric when the left side and right side have approximately the same shape

Example:

             █
           ███
         █████
       ███████
     █████████
   █████████████
-------------------
        Mean

Important point

For a perfectly symmetric unimodal distribution:

Mean = Median = Mode

Business example:

If employee performance scores are evenly distributed around the average, the distribution may be approximately symmetric

AI/ML example:

Symmetric features are often easier to analyze statistically than heavily skewed features
```

```python
import numpy as np                              #PythonImplementation
import matplotlib.pyplot as plt

data = np.random.normal(50, 10, 1000)

plt.hist(data, bins=30)
plt.title("Symmetric Distribution")
plt.show()
```

### Kurtosis

```python
Kurtosis describes the shape of the tails of a distribution, particularly how much extreme-value behavior it has compared with a normal distribution

In simple terms:

Kurtosis helps us understand whether a distribution has relatively light or heavy tails

There are three commonly discussed types.

1. Mesokurtic

Similar tail behavior to a normal distribution.

Normal-like

2. Leptokurtic

Heavier tails and more extreme observations

More extreme values

3. Platykurtic

Lighter tails and fewer extreme observations

Fewer extreme values

Business example:

Suppose a bank studies transaction amounts

High kurtosis may indicate that there are more extreme transactions than expected

AI/ML example:

Kurtosis can help identify features with unusually heavy tails and possible extreme observations
```

```python
import pandas as pd                           #PythonImplemntation

data = [10, 12, 13, 14, 15, 16, 17, 18, 30]

series = pd.Series(data)

print("Kurtosis:", series.kurt())
```

### Empirical Rule (68–95–99.7)



The Empirical Rule applies to a normal or approximately normal distribution.

It tells us how much data falls within 1, 2, and 3 standard deviations from the mean.

Rule
Range	      Approximate data
Mean ± 1 SD   	68%
Mean ± 2 SD  	95%
Mean ± 3 SD  	99.7%


For 1 Standard Deviation:

$$
\mu-\sigma \leq X \leq \mu+\sigma
$$

Approximately:

$$
68\%
$$

For 2 Standard Deviations:

$$
\mu-2\sigma \leq X \leq \mu+2\sigma
$$

Approximately:

$$
95\%
$$

For 3 Standard Deviations:

$$
\mu-3\sigma \leq X \leq \mu+3\sigma
$$

Approximately:

$$
99.7\%
$$

```python
Numerical example:

Suppose:

Mean = 100
Standard Deviation = 10
Within 1 standard deviation
100 - 10 = 90
100 + 10 = 110

Approximately 68% of values are between:

90 and 110
Within 2 standard deviations
100 - 20 = 80
100 + 20 = 120

Approximately 95% of values are between:

80 and 120
Within 3 standard deviations
100 - 30 = 70
100 + 30 = 130

Approximately 99.7% of values are between:

70 and 130

Business example

A company measures delivery time:

Mean = 50 minutes
SD = 5 minutes

Approximately 68% of deliveries would be between:

45 and 55 minutes

if delivery times are approximately normally distributed

AI/ML example:

The empirical rule can help identify unusually far-away observations in approximately normal data

For example, a value beyond approximately:

    μ ± 3σ

may deserve investigation as an unusual observation    
```

```python
mean = 100
std = 10

print("68% range:", mean - std, "to", mean + std)
print("95% range:", mean - 2 * std, "to", mean + 2 * std)
print("99.7% range:", mean - 3 * std, "to", mean + 3 * std)
```
