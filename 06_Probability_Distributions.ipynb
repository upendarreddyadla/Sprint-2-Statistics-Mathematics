### Bernoulli Distribution

A Bernoulli Distribution is used when there is only one trial and there are only two possible outcomes:

Success → 1

Failure → 0

Examples:

Pass / Fail

Yes / No

Buy / Not Buy

Defective / Not Defective

Click / No Click

Formula:

$$
P(X=x)=p^x(1-p)^{1-x}
$$

where:

$$
x\in\{0,1\}
$$

Where:

p = probability of success

1-p = probability of failure

x = outcome

Example:

Suppose the probability that a customer buys a product is 0.7.

Success = Buy
Failure = Don't Buy

Therefore:

$$ P(X=1)=0.7 $$ $$ P(X=0)=0.3 $$

Where is it used?

Customer purchase prediction

Loan approval

Medical testing

Quality control

Click prediction

Real-World Example:

A company sends an advertisement.

Customer clicks → 1

Customer doesn't click → 0

AI/ML Example:

A model predicts whether a customer will churn:

1 → Customer churns
0 → Customer does not churn

```python
import numpy as np                               #PythonVisualization
import matplotlib.pyplot as plt

p = 0.7

x = [0, 1]
probabilities = [1-p, p]

plt.bar(x, probabilities)

plt.title("Bernoulli Distribution")
plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.xticks([0, 1], ["Failure", "Success"])

plt.show()
```

### Binomial Distribution

The Binomial Distribution tells us the probability of getting a certain number of successes in a fixed number of independent trials

Conditions:

A binomial distribution has:

Fixed number of trials

Only two outcomes

Trials are independent

Probability of success remains constant

Formula:

$$
P(X=k)=
\binom{n}{k}
p^k(1-p)^{n-k}
$$

Where:

n = number of trials

k = number of successes

p = probability of success

1-p = probability of failure

Example:

A coin is tossed 5 times

Probability of Head:

$$ p=0.5 $$

Probability of getting exactly 3 heads:

$$ P(X=3) = \binom{5}{3}(0.5)^3(0.5)^2 $$ $$ =0.3125 $$

Where is it used?

Number of successful sales

Number of customers who respond

Quality control

A/B testing

Marketing campaigns

Real-World Example:

A company sends offers to 100 customers

Suppose each customer has a 20% probability of purchasing

We can calculate the probability that exactly 25 customers purchase

AI/ML Example:

Predicting the number of customers who will click an advertisement out of 1,000 customers

```python
import numpy as np                              #PythonVisualiZation
import matplotlib.pyplot as plt
from math import comb

n = 10
p = 0.5

x = np.arange(0, n + 1)

probabilities = [
    comb(n, k) * (p ** k) * ((1 - p) ** (n - k))
    for k in x
]

plt.bar(x, probabilities)

plt.title("Binomial Distribution")
plt.xlabel("Number of Successes")
plt.ylabel("Probability")

plt.show()
```

### Poisson Distribution


The Poisson Distribution is used to model the number of times an event occurs in a fixed interval of time or space

Examples:

1.Number of calls per hour

2.Number of customers arriving per minute

3.Number of website errors per day

4.Number of accidents per month

Formula:

$$
P(X=k)=\frac{e^{-\lambda}\lambda^k}{k!}
$$

Where:

λ = average number of events

k = number of events

e ≈ 2.718

Example:

A call center receives an average of 5 calls per minute.

What is the probability of receiving exactly 3 calls in one minute?

Here:

$$ \lambda=5 $$ $$ k=3 $$

Therefore:

$$ P(X=3)= \frac{e^{-5}5^3}{3!} $$

Approximately:

$$ P(X=3)\approx0.1404 $$

So there is approximately a 14.04% probability

Where is it used?

Call centers

Website traffic

Network failures

Hospital arrivals

Manufacturing defects

Real-World Example:

A restaurant receives an average of 10 customers every hour

Poisson Distribution can estimate the probability of receiving exactly 15 customers in the next hour

AI/ML Example:

Predicting:

Number of customer complaints

Number of system failures

Number of website requests

```python
import numpy as np                                  #PythonVisulization
import matplotlib.pyplot as plt
from math import factorial

lam = 5

x = np.arange(0, 16)

probabilities = [
    (np.exp(-lam) * lam**k) / factorial(k)
    for k in x
]

plt.bar(x, probabilities)

plt.title("Poisson Distribution")
plt.xlabel("Number of Events")
plt.ylabel("Probability")

plt.show()
```

### Uniform Distribution


In a Uniform Distribution, all values in a specified range have equal likelihood for the continuous case

For example, if a random number is selected between 0 and 10, every point in the interval has the same density

Formula

$$
f(x)=\frac{1}{b-a}
$$

where:

$$
a\leq x\leq b
$$

Where:

a = minimum value
b = maximum value

Example:

Suppose a random number is selected between 10 and 20.

Then:

$$ a=10 $$ $$ b=20 $$

Therefore:

$$ f(x)=\frac{1}{20-10} $$ $$ =\frac{1}{10} $$

Where is it used?

Random number generation

Simulation

Games

Sampling

Computer algorithms

Real-World Example:

Suppose a bus can arrive at any time between 8:00 AM and 8:30 AM with equal likelihood.

We can model the arrival time approximately using a uniform distribution if that equal-likelihood assumption is reasonable.

AI/ML Example:

Uniform random values are often used when:

Initializing parameters

Random sampling

Simulation

Testing algorithms

```python
import numpy as np                                   #PythonVisualization
import matplotlib.pyplot as plt

data = np.random.uniform(10, 20, 1000)

plt.hist(data, bins=20)

plt.title("Uniform Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```

### Normal Distribution


The Normal Distribution is one of the most important probability distributions in statistics

It has a characteristic bell-shaped curve

It is symmetric around its mean

Formula:

$$
f(x)=
\frac{1}{\sigma\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
$$

Where:

μ = mean

σ = standard deviation

π ≈ 3.14159

e ≈ 2.718

Example:

Suppose student marks have:

Mean = 60
Standard Deviation = 10

Most students will have marks around 60

The distribution will look approximately like a bell

Empirical Rule:

For approximately normal data:

### Empirical Rule

$$
\mu-\sigma \leq X \leq \mu+\sigma
$$

Approximately:

$$
68\%
$$

$$
\mu-2\sigma \leq X \leq \mu+2\sigma
$$

Approximately:

$$
95\%
$$

$$
\mu-3\sigma \leq X \leq \mu+3\sigma
$$

Approximately:

$$
99.7\%
$$


Where is it used?

Exam scores

Heights

Measurement errors

Statistical inference

Quality control

Real-World Example:

Adult heights in a population can often be approximately modeled by a normal distribution within a suitable population

AI/ML Example:

Normal distributions appear in:

Statistical modeling

Gaussian Naive Bayes

Error modeling

Feature standardization

Probabilistic machine learning

```python
import numpy as np                                 #PythonVisualization
import matplotlib.pyplot as plt

data = np.random.normal(50, 10, 1000)

plt.hist(data, bins=30)

plt.title("Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```

### Exponential Distribution

The Exponential Distribution is used to model the waiting time until the next event when events occur independently at a constant average rate

Formula

$$
f(x)=\lambda e^{-\lambda x}
$$

where:

$$
x\geq0
$$

Where:

λ = rate of events

x = waiting time

Example:

Suppose a call center receives an average of 2 calls per minute.

Then:

$$ \lambda=2 $$

The exponential distribution can model the waiting time until the next call.

Where is it used?

Waiting-time analysis

Reliability engineering

Queueing systems

Customer service

Network traffic

Real-World Example:

A customer arrives at a bank

We want to estimate how long they may wait until the next customer arrives

AI/ML Example:

Exponential distributions can be used in:

Survival analysis

Reliability modeling

Event-time modeling

Queueing models

```python
import numpy as np                              #PythonVisualization
import matplotlib.pyplot as plt

lam = 0.5

data = np.random.exponential(1 / lam, 1000)

plt.hist(data, bins=30)

plt.title("Exponential Distribution")
plt.xlabel("Waiting Time")
plt.ylabel("Frequency")

plt.show()
```
