## Outlier Detection

Outlier Detection is the process of finding unusually high or unusually low values in a dataset

In simple words:

An outlier is a data value that is very different from the other values

Example:

10, 12, 11, 13, 12, 14, 100

100 may be an outlier.

Statistics gives us several methods.

Method 1: IQR

Formula:

$$ IQR = Q3 - Q1 $$

Lower boundary:

$$ Q1 - 1.5(IQR) $$

Upper boundary:

Q3+1.5(IQR)

```python
import numpy as np

ages = [20, 22, 24, 25, 26, 27, 100]

Q1 = np.percentile(ages, 25)
Q3 = np.percentile(ages, 75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = [age for age in ages if age < lower or age > upper]

print("Outliers:", outliers)
```

```python
Simple Explanation:
    
Q1 → 25% point of the data.
Q3 → 75% point of the data.
IQR → Difference between Q3 and Q1.
lower → Lower limit.
upper → Upper limit.
Values outside these limits are outliers.
```

Method 2: Z-Score

$$ Z = \frac{x-\mu}{\sigma} $$

A very large absolute Z-score can indicate an unusual observation.

## Data Normalization

Data Normalization is the process of scaling numerical data to a common range, usually 0 to 1.

In simple words:

Normalization makes different numerical values comparable by bringing them into the same range.
Formula:

$$ x' = \frac{x-x_{min}}{x_{max}-x_{min}} $$

Usually the result is between:

0 and 1

Example:

Age = 30

Minimum age = 20

Maximum age = 40

$$ x' = \frac{30-20}{40-20} $$
x
′
=0.5

```python
from sklearn.preprocessing import MinMaxScaler

X = [[20], [25], [30], [35], [40]]

scaler = MinMaxScaler()

X_normalized = scaler.fit_transform(X)

print(X_normalized)
```

```python
Simple Explanation:
    
X → Original Age values.
MinMaxScaler() → Creates the normalization tool.
fit_transform() → Finds the minimum and maximum values and converts the data.
The values are changed to a range between 0 and 1.

Simple idea:

Original Data → Min-Max Scaling → Normalized Data (0 to 1)
```

## Linear Regression


Linear Regression is a supervised machine learning algorithm used to predict a continuous numerical value based on one or more input features

Example:

Predict house price using house size.

The basic equation is:

$$ y = b_0 + b_1x $$

Where:

\(y\) = predicted value
\(x\) = input feature
\(b_0\) = intercept
\(b_1\) = coefficient

For multiple features:

y = b0 + b1x1 + b2x2+⋯+bnxn


Where does statistics come in?

Linear regression uses statistical concepts such as:

Mean

Variance

Covariance

Correlation

Least squares

Residuals

R²

Confidence intervals

Hypothesis testing


```python
from sklearn.linear_model import LinearRegression

X = [[1], [2], [3], [4], [5]]

y = [40, 50, 60, 70, 80]

model = LinearRegression()

model.fit(X, y)

prediction = model.predict([[6]])

print("Predicted Score:", prediction)
```

```python

X → Study hours.
y → Exam scores.
LinearRegression() → Creates the model.
fit() → Model learns from the data.
predict() → Predicts a new value.
For 6 study hours, the model predicts approximately 90 marks.

Simple idea:

Study Hours → Linear Regression → Exam Score

Linear Regression is used to predict a continuous numerical value, such as salary, price, marks, or temperature.
```

## Logistic Regression

Logistic Regression is a machine learning algorithm used to predict categories or classes, usually Yes/No or 0/1.

In simple words:

Logistic Regression predicts the probability of an outcome and then assigns it to a category.

Example:

Will a customer buy a product?
Yes / No

It predicts probability.

$$ P(Y=1|x)=\frac{1}{1+e^{-(\beta_0+\beta_1x)}} $$

For example:

Probability = 0.82

We might classify:

0.82 >= 0.5 → Yes

Where does statistics come in?

Logistic regression uses:

Probability

Odds

Log-odds

Maximum likelihood

Coefficients

Hypothesis testing

Confidence intervals

Statistics → Probability → Classification

```python
from sklearn.linear_model import LogisticRegression

X = [[18], [20], [22], [25], [30], [35], [40], [45]]

y = [0, 0, 0, 0, 1, 1, 1, 1]

model = LogisticRegression()

model.fit(X, y)

prediction = model.predict([[28]])

print("Prediction:", prediction)
```

```python
This means:

1 = Yes

So the model predicts:

A 28-year-old is likely to buy the product

X → Age values.
y → Whether the customer buys the product.
0 → No.
1 → Yes.
```

## Naive Bayes

$$ P(A|B)=\frac{P(B|A)P(A)}{P(B)} $$

Naive Bayes is a machine learning algorithm used to classify data into categories using probability.

Simple Example:

Example: Predict whether an email is Spam or Not Spam.

Spam = 1
Not Spam = 0

The model looks at words in the email and calculates which category is more likely.

Simple idea:

Email Words → Probability → Naive Bayes → Spam / Not Spam

Real-Life Examples:

Naive Bayes is based on Bayes' Theorem and is especially useful for text classified

```python
from sklearn.naive_bayes import GaussianNB

X = [[1], [2], [3], [8], [9], [10]]

y = [0, 0, 0, 1, 1, 1]

model = GaussianNB()

model.fit(X, y)

prediction = model.predict([[7]])

print("Prediction:", prediction)
```

## Clustering

```python
clustering is an unsupervised machine learning technique that groups similar data points together

Simple Example

Suppose we have customers based on their Age and Spending:

Customer	Age	 Spending
   A	     20	   100
   B	     22	   120
   C	     25	   110
   D	     40	   500
   E	     42	   520
   F	     45	   480

Clustering can create groups such as:

Group 1: Younger customers with low spending
Group 2: Older customers with high spending

Simple Real-Life Idea

Customers → Find Similar Behavior → Create Groups → Target Each Group
```

```python
from sklearn.cluster import KMeans

X = [
    [20, 100],
    [22, 120],
    [25, 110],
    [40, 500],
    [42, 520],
    [45, 480]
]
model = KMeans(n_clusters=2, random_state=42)

model.fit(X)

print(model.labels_)
```

```python
Simple Explanation:
    
X → Customer data.
KMeans → Clustering algorithm.
n_clusters=2 → Create 2 groups.
fit() → Finds similar customers and creates groups.
labels_ → Shows which group each customer belongs to.

Simple idea:

Customer Data → Find Similar Customers → Create Groups → Cluster
```

## Principal Component Analysis (PCA)

```python
Principal Component Analysis (PCA) is a technique used to reduce the number of features in a dataset while keeping the most important information

In simple words:

PCA converts many features into fewer important features.

For example:

Original Data

Age + Salary + Experience + Spending
              ↓
             PCA
              ↓
       Component 1
       Component 2

Real-Time Example:

Customer Data:

A shopping company may have many customer features:

Age
Income
Spending
Number of purchases
Website visits
Time spent on website

Instead of using all these features, PCA can reduce them to a smaller number of principal components while preserving much of the useful variation.

This can make data easier to visualize, process, and use in machine learning models

3 Features
   ↓
   PCA
   ↓
2 Important Components    
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

X = [
    [20, 20000, 1],
    [25, 30000, 2],
    [30, 40000, 4],
    [35, 50000, 6],
    [40, 60000, 8]
]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print(X_pca)
```

```python
Explanation:
    
X → Original data with 3 features
StandardScaler() → Puts features on a similar scale
PCA(n_components=2) → Reduces 3 features to 2 components
fit_transform() → Learns the important patterns and transforms the data
X_pca → New reduced dataset
```
