### Mean

```python
Mean is the average value of a set of numbers
Mean = Sum of all values ÷ Number of values

Formula:        
                           
	      Number of values
    Mean= ---------------
          Sum of all values

Easy way to remember

Mean = Average

Example:- Numercal                                                #Numerical Example

Suppose the marks are:

10, 20, 30, 40, 50

Step 1: Add all values

10 + 20 + 30 + 40 + 50 = 150

Step 2: Count the values

There are 5 values

Step 3: Divide

Mean = 150 ÷ 5 = 30

So, the Mean = 30

Business Example:                                                   #Business Example

A company has monthly sales:

₹10,000, ₹20,000, ₹30,000, ₹40,000, ₹50,000

The mean sales are:

₹30,000

This tells the company its average monthly sales


AI/ML Example:                                                   #AIML Example                             

Suppose a dataset contains customer ages:

20, 25, 30, 35, 40

Mean age = 30

The mean can be used for data analysis and preprocessing

Interpretation:                                                   #Interpretation

A mean of 30 means:

The dataset has an average value of 30.

Important: Mean can be strongly affected by extreme values (outliers)    
```

```python
data = [10, 20, 30, 40, 50]                    #Python Implementation of Mean

mean = sum(data) / len(data)

print("Mean:", mean)
```

### Median

For \(n\) values:

- If \(n\) is odd:
$$
\text{Median} = X_{\frac{n+1}{2}}
$$

- If \(n\) is even:
$$
\text{Median} = \frac{X_{\frac{n}{2}} + X_{\frac{n}{2}+1}}{2}
$$


```python
Median is the middle value in a dataset when the values are arranged in ascending or descending order
Median = The middle value of ordered data

Odd → One middle value
Even → Average of two middle values

Example:

Data:

10, 20, 30, 40, 50

There are 5 values

The middle value is 30.

So,The Median is = 30

Example 2: Even number of values                              #Numerical Example

Data:

10, 20, 30, 40

There are 4 values, so there is no single middle value

Take the two middle values:

20 and 30

Then calculate their average:

        
          20 + 30
 Median= ---------
            2

 Median = 25

Steps to find Median:
     
1.Arrange the data in ascending order
2.Count the number of values
3.If the number is odd → choose the middle value
4.If the number is even → take the average of the two middle values

Business Example:                                                     #Business Example

Consider employee salaries:

₹20,000, ₹25,000, ₹30,000, ₹35,000, ₹5,00,000

The mean is heavily affected by the ₹5 lakh salary

The median is ₹30,000

Therefore, median gives a better idea of the typical employee salary in this situation 

AI/ML Example:

Suppose a dataset contains customer transaction amounts:

100, 120, 150, 200, 10,000

The mean becomes very large because of the ₹10,000 transaction.

Median is much less affected by this outlier.

Median can therefore be useful when preparing data for ML models, especially when distributions are skewed    

Interpretation:-                                                          #Interpretation

Median tells us the central position of the data.

If:

Median = 30

it means approximately half the observations are below 30 and half are above 30
```

```python
import statistics                          #Python Implementation

numbers = [10, 20, 30, 40, 50]

median = statistics.median(numbers)

print("Median:", median)
```

### Mode


Mode is the value that occurs most frequently in a dataset
Mode = Most repeated value

Formula:

Mode = Most frequently occurring value

Example:                                                #Numerical Example

Data:

10, 20, 30, 30, 40, 20, 30

Here:

30 occurs 3 times

Therefore:

Mode = 30

Business Example:                                    #BusinessExample

A clothing store records T-shirt sizes sold:

M, L, L, S, M, L, XL, L

The most frequently sold size is:

Mode = L

The company can use this information to decide which size to keep in greater quantity

```python
import statistics

numbers = [10, 40, 20, 30, 40, 20, 40]

mode = statistics.mode(numbers)

print("Mode:", mode)
```

### Weighted Mean

#Weighted Mean Formula
$$
\text{Weighted Mean} = \frac{\sum (x_i w_i)}{\sum w_i}
$$

Where:

(x_i) = value
(w_i) = weight

```python
Weighted Mean is an average where different values have different levels of importance

In a normal mean, every value has equal importance

In a weighted mean, some values have more importance than others

Suppose your marks are:                                  #NumericalExample

Subject	  Marks	 Weight
Python	    80	   3
SQL      	70	   2
Statistics	90	   5

Calculate:

(80×3)+(70×2)+(90×5)
 =240+140+450=830

Total weight:

3+2+5=10

Therefore:
                  830
Weighted Mean = --------  = 83
                   10

Weighted Mean = 83

Business Example:                              #Business example

A company evaluates an employee using:

Project performance = 80, weight = 50%
Communication = 70, weight = 20%
Technical skills = 90, weight = 30%

The weighted average gives more importance to project performance

his is more appropriate than simply calculating:

      80+70+90
      --------
          3
because all criteria are not equally important.

AI/ML Example:                                            #AIML Example

Suppose a model's performance is evaluated using:

Accuracy = 90%, weight = 50%
Precision = 80%, weight = 30%
Recall = 70%, weight = 20%

          (90×0.5)+(80×0.3)+(70×0.2)

             =45+24+14=83

Weighted score = 83%

Interpretation:                                              #Interpretation

Weighted mean = 83 means the overall average is 83 after considering the importance of each value.

Normal Mean → Equal importance
Weighted Mean → Different importance
```

### Geometric Mean

For \(n\) values:

$$
\text{Geometric Mean} = \sqrt[n]{x_1 \cdot x_2 \cdot x_3 \cdot \dots \cdot x_n}
$$


```python
Geometric Mean is an average calculated by multiplying all values and then taking the \(n\)-th root.

It is particularly useful for growth rates, returns, ratios, and percentages

Numerical Example

Suppose we have:

2, 8

Step 1: Multiply:

    2×8=16

Step 2: Take the square root:

Therefore:

Geometric Mean = 4

Business Example:                                                 #BusinessExample

Suppose a company's investment grows by:

Year 1: 10%
Year 2: 20%
Year 3: 30%

For growth rates, the geometric mean is often more appropriate than the simple mean because growth compounds over time

A business can use geometric mean to understand the average compound growth rate

AI/ML Example:                                                  #AIMLExample

Suppose different multiplicative factors are:

1.1, 1.2, 1.3

The geometric mean can be used to summarize the typical multiplicative growth factor

It is useful in areas involving ratios, rates, and multiplicative effects

Interpretation:                                                    #Interpretation

Geometric mean represents an average multiplicative rate

It is especially useful when values are being multiplied or compounded

Arithmetic Mean → Addition
Geometric Mean → Multiplication / Growth
```

### Harmonic Mean

For \(n\) values:

$$
\text{Harmonic Mean} = \frac{n}{\frac{1}{x_1} + \frac{1}{x_2} + \cdots + \frac{1}{x_n}}
$$


```python
Harmonic Mean is an average based on the reciprocals of values

It is especially useful for rates and ratios, such as speed

Numerical Example:                                #NumericalExample

Suppose a car travels:

First half of a journey at 60 km/h
Second half at 40 km/h

Because the distances are equal, we use the harmonic mean
                          2
              H = ----------------------
                     1          1
                    ----  +  -------
                     60          40

Calculate:
         2
H = -------------
    0.01667 + 0.025

	

  H = 48

Therefore:

Average speed = 48 km/h

Business Example:                          #BusinessExample

Suppose a company measures:

60 units/hour in one process
40 units/hour in another process

When combining rates under appropriate conditions, the harmonic mean can provide a meaningful average rate

It is also commonly relevant when averaging ratios or rates

AI/ML Example:                           #AIMLExample

The most important ML example is the F1-score.

F1-score is the harmonic mean of:

Precision
Recall

         F1 = 2 * Precision + Recall 
              --------------------
               Precision + Recall

   Suppose:

Precision = 0.80
Recall = 0.60
       
          2(0.8)(0.6)
	F1 = -------------
           0.8 + 0.6

  F1 ≈ 0.686

F1-score ≈ 0.686 or 68.6%

Interpretation:
                

Harmonic mean is useful when dealing with rates, ratios, and situations where low values should strongly affect the overall average.

In ML:

High precision + high recall → High F1-score

If either precision or recall is low, the F1-score will also be relatively low       
```

```python
from statistics import harmonic_mean                #PythonImplementation

speeds = [60, 40]

result = harmonic_mean(speeds)

print("Harmonic Mean:", result)
```
