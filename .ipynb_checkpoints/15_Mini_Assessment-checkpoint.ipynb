## Baye's Theorem

### Coding Exercise

Write a Python program to calculate Bayes' Theorem

```python
P_A = 0.2
P_B_given_A = 0.5
P_B = 0.4

result = (P_B_given_A * P_A) / P_B

print("P(A|B):", result)
```

## Correlation and Covariance

### Coding Exercise

Write a Python program to calculate Correlation and Covariance using NumPy

```python
import numpy as np

x = [10, 20, 30, 40, 50]
y = [20, 30, 40, 50, 60]

correlation = np.corrcoef(x, y)[0, 1]
covariance = np.cov(x, y, ddof=0)[0, 1]

print("Correlation:", correlation)
print("Covariance:", covariance)
```

 ## IQR

### Coding Exercise

Write a Python program to calculate the IQR (Interquartile Range) of a dataset using NumPy

```python
import numpy as np

data = [10, 20, 30, 40, 50, 60, 70]

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

print("IQR:", IQR)
```

##  IQR Method

### Coding Excercise

```python
import pandas as pd

data = pd.Series([10, 12, 15, 14, 13, 150])

Q1 = data.quantile(0.25)
Q3 = data.quantile(0.75)

IQR = Q3 - Q1

outliers = data[(data < Q1 - 1.5 * IQR) | (data > Q3 + 1.5 * IQR)]

print("Outliers:", list(outliers))
```

## Correlation between two variables


### Coding Excercise

Write Python code to calculate the Correlation between two variables

```python
import pandas as pd

x = [10, 20, 30, 40, 50]
y = [20, 30, 40, 50, 60]

data = pd.DataFrame({"X": x, "Y": y})

correlation = data["X"].corr(data["Y"])

print("Correlation:", correlation)
```
