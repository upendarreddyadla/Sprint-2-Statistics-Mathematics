### Introduction to Probability

Probability is the measure of how likely an event is to happen.

The value of probability is always between 0 and 1.

0 → Impossible
0.5 → Equal chance
1 → Certain

Formula
### Probability Formula

$$
P(A)=\frac{\text{Number of Favorable Outcomes}}
{\text{Total Number of Outcomes}}
$$

Example:

Suppose we toss a fair coin.

Possible outcomes:

{Head, Tail}

Probability of getting Head:

$$ P(H)=\frac{1}{2}=0.5 $$


Business Example:

A company may estimate:

Probability that a customer will buy a product = 0.7

This means there is approximately a 70% chance of purchase

AI/ML Example:

A classification model may predict:

Customer will churn: 0.80
Customer will not churn: 0.20

The model estimates an 80% probability of churn

```python
favorable = 1                             #PythonImplementation
total = 2

probability = favorable / total

print("Probability:", probability)
```

### Events

An event is an outcome or collection of outcomes that we are interested in.

Example

When rolling a die:

Sample Space = {1, 2, 3, 4, 5, 6}

Event A = getting an even number:

A = {2, 4, 6}

Therefore:
        3       1
P(A) = ----- = ----
         6       2


```python
outcomes = [1, 2, 3, 4, 5, 6]             #PythonImplementation

even_numbers = [2, 4, 6]

probability = len(even_numbers) / len(outcomes)

print("Probability of even number:", probability)
```

### Sample Space

```python
The sample space is the set of all possible outcomes of an experiment

Example 1 — Coin

S = {Head, Tail}

Example 2 — Die

S = {1, 2, 3, 4, 5, 6}

Example 3 — Two Coins

S = {HH, HT, TH, TT}

There are 4 possible outcomes
```

```python
sample_space = ["HH", "HT", "TH", "TT"]                   #PythonImplementation

print("Sample Space:", sample_space)
print("Total outcomes:", len(sample_space))
```

### Independent Events

### Independent Events

Two events are dependent when the occurrence of one event affects the probability of the other event.

Example

Suppose a bag contains:

3 Red balls
2 Blue balls

You pick one ball and do not replace it.

The probability of the second selection changes because the first ball was removed.

$$
P(A\cap B)=P(A)\times P(B)
$$

```python
p_A = 1/2                 #PythonImplementation
p_B = 1/2

probability = p_A * p_B

print("Probability:", probability)
```

### Dependent Events

Two events are dependent when the occurrence of one event affects the probability of the other event.

Example

Suppose a bag contains:

3 Red balls
2 Blue balls

You pick one ball and do not replace it.

The probability of the second selection changes because the first ball was removed.

Formula
### Dependent Events

$$
P(A\cap B)=P(A)\times P(B|A)
$$

Example:

Probability of selecting two red balls:

First red:

$$ P(A)=\frac35 $$

After selecting one red ball:

Red = 2
Total = 4

Therefore:

$$ P(B|A)=\frac24 $$

So:


$$
P(H_1 \cap H_2)
=
\frac{1}{2}\times\frac{1}{2}
=
\frac{1}{4}
$$

```python
p_first_red = 3/5                     #PythonImplementation
p_second_red = 2/4

probability = p_first_red * p_second_red

print("Probability:", probability)
```

### Conditional Probability

Conditional probability is the probability of event A happening given that event B has already happened.

It is written as:

P(A∣B)

Formula


$$
P(A|B)=\frac{P(A\cap B)}{P(B)}
$$

Example:

Suppose:

$$ P(A\cap B)=0.30 $$

and:

$$ P(B)=0.50 $$

Then:

P(A∣B)=
0.50
0.30
	​

=0.60

So the conditional probability is 60%

AI/ML Example:

Suppose:

A = customer will churn
B = customer contacted support

We can calculate:

$$ P(\text{Churn}|\text{Support Contact}) $$

This tells us the probability that a customer churns given that they contacted support

```python
p_A_and_B = 0.30                                #PythonImplementation
p_B = 0.50

conditional_probability = p_A_and_B / p_B

print("P(A|B):", conditional_probability)
```

### Bayes' Theorem

Bayes' Theorem helps us update the probability of an event when we receive new evidence


$$
P(A|B)=
\frac{P(B|A)P(A)}
{P(B)}
$$

Where:

P(A|B) = probability of A given B
P(B|A) = probability of B given A
P(A) = prior probability
P(B) = probability of evidence

 Example:

Suppose:

P(A) = 0.20
P(B|A) = 0.80
P(B) = 0.40

Then:

$$ P(A|B) = \frac{0.80\times0.20}{0.40} $$ $$ =0.40 $$

Therefore:

P(A|B) = 40%

AI/ML Example:

Bayes' theorem is important in:

Spam detection
Medical diagnosis
Fraud detection
Naive Bayes classification
Document classification

For example:

Email contains suspicious words
          ↓
Update probability
          ↓
Probability of Spam          

```python
p_A = 0.20                         #PythonImplementation
p_B_given_A = 0.80
p_B = 0.40

p_A_given_B = (p_B_given_A * p_A) / p_B

print("P(A|B):", p_A_given_B)
```

### Probability Rules

There are several important probability rules

Rule 1 — Complement Rule

The probability that an event does not happen is:

### Complement Rule

$$
P(A^c)=1-P(A)
$$
Example

If:

$$ P(A)=0.7 $$

Then:

$$ P(A^c)=1-0.7=0.3 $$

So there is a 30% probability that A does not happen

Rule 2 — Addition Rule

For two events:

$$
P(A\cup B)=P(A)+P(B)-P(A\cap B)
$$

Example:

Suppose:

$$ P(A)=0.5 $$ $$ P(B)=0.4 $$ $$ P(A\cap B)=0.2 $$

Then:

$$ P(A\cup B) = 0.5+0.4-0.2 $$ $$ =0.7 $$

Rule 3 — Multiplication Rule


$$
P(A\cap B)=P(A)\times P(B|A)
$$

For independent events:

$$ P(A\cap B)=P(A)P(B) $$


### Random Variables

A random variable is a variable whose value depends on the outcome of a random experiment.

Usually represented by:

$$ X $$

Example:

Toss a coin

Define:

X = number of Heads

If we toss the coin twice:

HH → X = 2

HT → X = 1

TH → X = 1

TT → X = 0

So possible values are:

X∈{0,1,2}

Types of Random Variables

1.Discrete Random Variable

Takes countable values

Examples:

Number of customers
Number of defective products
Number of calls

2.Continuous Random Variable

Can take values within a range

Examples:

Height
Weight
Temperature
Time

```python
outcomes = ["HH", "HT", "TH", "TT" ]                       #PythonImplementation

for outcome in outcomes:
    heads = outcome.count("H")
    print(outcome, "→ X =", heads)
```

### Expected Value

Expected Value is the long-run average value we expect from a random variable

It is also called the mean of a probability distribution


$$
E(X)=\sum_x xP(X=x)
$$

Simple Example:

Suppose a game has:

Outcome	Probability
₹100	0.5
₹200	0.3
₹0	0.2

Expected value:

$$ E(X) = 100(0.5)+200(0.3)+0(0.2) $$ $$ =50+60+0 $$
=₹110

Business Example:

A company can calculate expected revenue from different possible sales outcomes.

For example:

High sales     → ₹1,00,000
Medium sales   → ₹60,000
Low sales      → ₹20,000

Each outcome has a probability. Expected value gives the probability-weighted average revenue.

AI/ML Example:

Expected value is used in:

Decision making
Risk analysis
Reinforcement learning
Probabilistic Models
Expected loss Calculations

```python
outcomes = [100, 200, 0]                   #PythonImplementation
probabilities = [0.5, 0.3, 0.2]

expected_value = sum(
    x * p for x, p in zip(outcomes, probabilities)
)

print("Expected Value:", expected_value)
```
