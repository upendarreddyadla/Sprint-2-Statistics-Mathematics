## Notebook 1 – Introduction to Statistics

### What is Statistics?

```python
Statistics is the process of collecting, organizing, analyzing, interpreting, and presenting data to understand information and make decision

Main Steps of Statistics
Collect Data → Gather information
Organize Data → Arrange the information
Analyze Data → Calculate useful values
Interpret Data → Understand what the results mean
Present Data → Show results using tables, charts, or graphs

Example:

Suppose 5 students scored:

60, 70, 80, 90, 100

Using statistics, we can calculate:

Average (Mean) = 80
Highest score = 100
Lowest score = 60
Number of students = 5

So, statistics helps us understand the data instead of just looking at individual numbers
```

### Why Statistics is important in AI/ML

```python
Statistics is important in Artificial Intelligence and Machine Learning because it helps us understand data, 
find patterns, make predictions, and measure how accurate a model is

Statistics is the foundation that helps AI/ML understand data and make reliable predictions    

1.Understanding Data

Statistics helps us understand the basic characteristics of a dataset

For example, we can calculate:

Mean – average value
Median – middle value
Mode – most frequently occurring value
Minimum & Maximum
Standard Deviation – how spread out the data is

2.Finding Patterns

ML models learn patterns from data

Example:

If we have:

Hours Studied	Exam Score
   2	             40
   4              	 55
   6	             70
   8	             85

Statistics can help us identify that as study hours increase, exam scores tend to increase

3.Handling Uncertainty

Real-world data is not always perfect. Statistics helps us understand uncertainty and variation in data

For example:

A model predicts that a customer has a 75% probability of buying a product
Statistics helps us understand what that probability means

4.Data Preprocessing

Statistics is useful for identifying:

Missing values
Outliers
Unusual data
Data distribution

For example, if most salaries are between ₹20,000–₹80,000 but one value is ₹10,00,000, statistics can help identify it as a possible outlier

5.Feature Selection

Statistics can help determine which variables are useful for making predictions.

Example: To predict house prices, we might have:

House size 
Number of bedrooms 
Location 

6.Evaluating ML Models

After training a model, we need to know how well it performs.

Statistics helps with concepts such as:

Accuracy
Precision
Recall
Mean Squared Error (MSE)
Confidence intervals
Hypothesis testing

7.Avoiding Overfitting

Statistics helps us understand whether a model has learned general patterns or simply memorized the training data

8.Making Predictions

Many ML algorithms are based on statistical concepts.

Examples include:

Linear Regression
Logistic Regression
Naive Bayes
Decision Trees
Bayesian methods
```

### Statistics vs Machine Learning


```python
Statistics and Machine Learning (ML) both work with data, but their main goals are different

Definition:

Statistics-Focuses on understanding data and relationships

Machine Learning-Focuses on learning patterns from data to make predictions or decisions

Statistics = Understand the data
Machine Learning = Learn from the data and predict

Comparison:

Statistics                                                         Machine Learning
                                                  
1.Mainly focuses on understanding data                       	1.Mainly focuses on predictions/decisions
2.Uses mathematical and statistical methods                 	2.Uses algorithms to learn patterns
3.Often asks "Why?" or "What is the relationship?"	            3.Often asks "What will happen next?"
4.Can work well with smaller datasets	                        4.Often benefits from larger datasets
5.Emphasizes interpretation	                                    5.Often emphasizes predictive performance
Examples: Mean, Median, Hypothesis Testing              	    Examples: Decision Tree, Random Forest, Neural Network


Example:

Suppose we have this data:

Hours Studied	Exam Score
  2               	40
  4	                55
  6	                70
  8	                85

Statistics:

We might ask:

"Is there a relationship between study hours and exam scores?"

We can use correlation or regression to analyze the relationship

Machine Learning:

We might ask:

"If a student studies for 10 hours, what score might they get?"

We can train a model to make that prediction

How they work together

Data
  ↓
Statistics
  ↓
Understand & Analyze Data
  ↓
Machine Learning
  ↓
Learn Patterns
  ↓
Prediction / Decision
```

### Descriptive Statistics

```python
Descriptive Statistics is a branch of statistics that is used to summarize, organize, and describe data in a simple and understandable way
Descriptive Statistics = Describing what the data looks like

Example:

Suppose the marks of 5 students are:

60, 70, 80, 80, 90

Using descriptive statistics, we can find:

Mean = 76
Median = 80
Mode = 80
Minimum = 60
Maximum = 90
Range = 90 − 60 = 30

These values help us quickly understand the dataset

Important Types of Descriptive Statistics:-

1.Measures of Central Tendency

These tell us the central or typical value of the data

Mean → Average
Median → Middle value
Mode → Most frequently occurring value

Example:

10, 20, 20, 30, 40

Mean = 24
Median = 20
Mode = 20

2.Measures of Dispersion

These tell us how spread out the data is.

Common measures:

Range
Variance
Standard Deviation
Interquartile Range (IQR)

For example, if two classes have the same average marks, dispersion can tell us whether the students' marks are closely grouped or widely spread

3.Frequency

Frequency tells us how many times a value occurs

Example:

Apple, Banana, Apple, Orange, Apple

Frequency:

 Fruit	  Frequency
 Apple	    3
 Banana   	1
 Orange  	1

4.Data Visualization

Descriptive statistics can also use charts and graphs to summarize data:

1.Bar Chart
2.Histogram
3.Pie Chart
4.Box Plot
5.Line Chart
```

### Inferential Statistics

```python
Inferential Statistics is a branch of statistics that uses sample data to make conclusions, estimates, or predictions about a larger population
Inferential Statistics = Using a sample to make conclusions about a population

Example:

Imagine a company has 10,000 customers

You cannot ask every customer whether they like a new product

Instead, you select 500 customers and ask them

Suppose:

400 out of 500 customers like the product
That means 80% of the sample likes it

Inferential statistics can be used to estimate whether a similar proportion of the 10,000 customers might like the product
Population
10,000 customers
       ↓
Select Sample
500 customers
       ↓
Analyze Sample
80% like the product
       ↓
Make Inference
Estimate for the population

Main Uses of Inferential Statistics:
    
1.Estimation

We use a sample to estimate a population value

Example:

100 students average height = 168 cm

We may use this to estimate the average height of all students

2.Hypothesis Testing

We test whether there is enough evidence to support a claim

Example:

"Does a new teaching method improve student marks?"

We collect sample data and perform a statistical test

3.Confidence Intervals

A confidence interval gives a range of plausible values for a population parameter

Example:

Estimated average salary = ₹30,000

95% confidence interval:
₹28,000 to ₹32,000

The interval communicates uncertainty in the estimate

4.Prediction

Inferential statistics can also be used to make predictions based on observed data

For example:

Previous sales data
       ↓
Statistical analysis
       ↓
Estimate future sales
```

### Population

```python
In statistics, Population means the complete group of people, objects, or observations that we want to study or understand
Population = The entire group that we are interested in studying

Example: 
Students

Suppose a college has 5,000 students

If we want to study the average height of all students:

Population = All 5,000 students
Population size = 5,000

We may select only 500 students to collect data. Those 500 students are called a sample.

    Population:

  5,000 Students
      ↓
   Select
      ↓
    Sample
  500 Students
      ↓
  Analyze
      ↓
 Make conclusions about Population
```

### Sample

```python
A sample is a smaller group selected from a larger population for the purpose of collecting and analyzing data

Population = Whole group
Sample = Part of the whole group

Example:

Suppose a college has 5,000 students

You want to find the average height of the students

It may be difficult to measure all 5,000 students, so you select 100 students

              Population
  5,000 Students
       ↓
 Select 100 Students
       ↓
   Sample
 100 Students
       ↓
 Analyze their heights
       ↓
 Infer about the population

Here:

Population = 5,000 students
Sample = 100 students
Sample size = 100
```

### Census

```python
A census in statistics is the process of collecting data from every member of a population. Unlike sampling, which studies only a part of the population,
a census attempts to gather complete information about the entire group

Covers the whole population

Provides the most accurate and comprehensive data

Usually conducted by governments or large organizations because it requires significant time, money, and effort

Used as a baseline for planning, policymaking, and statistical analysis

Example:

National Census: The Government of India conducts a census every 10 years to collect data on population size, literacy rates, employment, housing..
```

### Sampling


```python
Sampling is the process of selecting a subset of individuals or observations from a larger population to represent the whole.
Since studying an entire population (census) is often expensive and time-consuming,
sampling provides a practical way to gather insights

Used when a census is not feasible

Must be representative of the population to avoid bias

Helps in making inferences about the population using inferential statistics

#Example:
Election Polls: Instead of asking all voters, a survey may sample 2,000 voters to predict the election outcome
```

### Variables

```python
A variable is any characteristic or value that can differ from one person, object, or observation to another

E:-xampleif you collect information about students, variables could include age, height, gender, test score, and favorite subject.

The main types of variables are:

Qualitative (categorical) variables — describe categories or qualities, such as blood type, eye color, or marital status
Nominal: categories with no natural order, e.g. blood type
Ordinal: categories with an order, e.g. poor, fair, good, excellent

Quantitative (numerical) variables — represent numbers or amounts
Discrete: countable values, usually whole numbers, e.g. number of children
Continuous: can take any value within a range, e.g. height, weight, or temperature
```

### Features

```python
A feature is an individual piece of information (variable) that describes something and can be used for analysis or prediction
Feature = An input variable used to describe or predict something

Example:

Suppose we want to predict a student's exam score:

Student	Age	StudyHours	Attendance	Exam Score
 A	    20 	   5	       90%	        85
 B	    21 	   3	       75%	        70
 C	    20	   7           95%	        92

Here:

Age → Feature
Study Hours → Feature
Attendance → Feature
Exam Score → Target/Output

Features → Input information
Target → What we want to predict

Real-life example

For predicting house prices:

House size → Feature
Number of bedrooms → Feature
Location → Feature
Age of house → Feature
House price → Target
Easy way to remember

Feature = Input
Target = Output

In Machine Learning, features are also commonly called independent variables or predictor variables
```

### Observation

```python
An observation is one individual data value or one complete record collected for a person, object, or event
Observation = One record of data collected from one individual or item

Example:

Suppose we collect information about 3 students:

Student	Age	StudyHours	Marks
 A	    20	    5	     80
 B	    21	    3	     65
 C	    19      6	     90

Here:

Student A's complete row → One observation
Student B's complete row → One observation
Student C's complete row → One observation

So,there are three observations
```

### Dataset

```python
A dataset is a collection of related data organized together for analysis
Dataset = A collection of observations and variables/features

Example:

Suppose we have student information:

Student	Age	StudyHours	Marks
 Ajay	20	   5	      80
 Vijay	21	   3	      65
 Vicky	19	   6	      90

This complete table is called a dataset

Dataset → Complete table
Observation → One row
Feature/Variable → One column
Value → One individual data item

Ex:-
Ajay | 20 | 5 | 80 → One observation

Age-One feature/variable

20-One value

Easy way to remember

Dataset = Rows + Columns + Values

Example:

3 rows × 4 columns = 3 observations and 4 variables
```

### Quantitative Data

```python
Quantitative data is data that can be measured or counted and represented using numbers
Quantitative Data = Numerical data that tells us "how much" or "how many"

Examples:

Age → 25 years
Height → 5.8 feet
Weight → 63 kg
Salary → ₹30,000
Number of students → 50
Exam marks → 85

Types of Quantitative Data

There are 2 main types:

1.Discrete Data:
Data that can be counted
Usually whole numbers
Example: Number of students = 40

2.Continuous Data:
Data that can be measured
Can have decimal values
Example: Height = 5.8 feet, Weight = 63.5 kg

Easy way to remember:
Quantitative = Quantity = Numbers

Discrete → Count
Continuous → Measure
```

### Qualitative Data

```python
Qualitative data is data that describes qualities, characteristics, or categories rather than amounts or numbers
Qualitative Data = Non-numerical data that describes "what type" or "which category"

Examples:

Gender → Male / Female
Eye color → Brown / Blue / Black
Blood group → A / B / AB / O
City → Hyderabad / Chennai / Delhi
Product type → Mobile / Laptop / Tablet
Customer feedback → Good / Average / Poor

Types of Qualitative Data

There are mainly 2 types:

1.Nominal Data
Categories with no specific order
Example: Eye color → Black, Brown, Blue

2.Ordinal Data
Categories that have a meaningful order
Example: Customer satisfaction → Poor → Average → Good → Excellent

Easy way to remember:

Quantitative → Numbers → "How much/how many?"
Qualitative → Categories/qualities → "What type/which?"
```

### Continuous Data

```python
Continuous data is a type of quantitative data that can take any value within a range. It is usually obtained by measuring something

Continuous Data = Data that can have decimal values and is measured rather than counted

Examples:

Height → 5.5 ft, 5.55 ft, 5.555 ft
Weight → 63 kg, 63.5 kg, 63.75 kg
Temperature → 36.5°C
Time → 10.25 seconds
Distance → 5.7 km

Example:

Suppose we measure the heights of students:

Student	Height
Ajay	5.5 ft
Vijay	5.2 ft
Vicky	5.75 ft

Height is continuous data because it can take many possible values, including decimals
Continuous → Measure → Decimal values possible
```

### Discrete Data

```python
Discrete data is a type of quantitative data that represents things we can count. It usually consists of whole numbers and cannot normally have meaningful decimal values
Discrete Data = Data that can be counted and usually has separate, distinct values

Examples:

Number of students → 50
Number of cars → 10
Number of children → 2
Number of employees → 100
Number of books → 25
Number of calls received → 15

For example, you can have 5 students or 6 students, but not 5.5 students

Discrete → Count
```

### Structured Data

```python
Structured data is data that is organized in a fixed format, usually in rows and columns, making it easy for computers to store, search, and analyz
Structured Data = Data organized in a predefined structure such as a table
                                                                                                         
Example:

Imagine a student database:

Student_ID	Name	Age	 Marks
 101         Ajay	 20	  85
102	         Vijay   21	  90
103          Vicky	 19	  75

This is structured data because:

Each column has a specific meaning
Each row represents one observation
Data follows a consistent format
It can easily be stored in a database or Excel table

Real-life examples:
    
Customer database
Employee records
Bank transactions
Student records   
```

### Unstructured Data

```python
Unstructured data is data that does not follow a fixed format or predefined structure like rows and columns
Unstructured Data = Data without a fixed or organized tabular structure

Examples:
Images → Photos, X-rays
Videos → YouTube videos, movies
Audio → Songs, voice recordings
Text → Emails, social media posts, reviews
Documents → PDF files, Word documents
Chat messages → WhatsApp or other messages

Example:

A customer writes:

"I really like this phone. The camera is excellent, but the battery could be better."

This is unstructured data because the information isn't arranged into fixed columns such as Product, Rating, Camera, and Battery
```
