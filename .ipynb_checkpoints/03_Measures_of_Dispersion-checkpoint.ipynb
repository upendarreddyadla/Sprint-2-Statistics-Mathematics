# Notebook 3 – Measures of Dispersion

```python
Measures of Dispersion tell us how much the data values are spread out or scattered around the center

For example:

Dataset A: 48, 49, 50, 51, 52
Dataset B: 10, 30, 50, 70, 90

Both datasets have:

$$ Mean = 50 $$

But Dataset B is much more spread out.

So:

Central Tendency → Where is the center?
Dispersion → How spread out is the data?
```

### Range

```python
Range is the difference between the maximum value and minimum value

Formula:

Range = Maximum−Minimum

Numerical Example

Data:

10, 20, 30, 40, 50

Maximum = 50
Minimum = 10

Range = 50 - 10 = 40 

Range = 40

Business Example:                                           #BusinessExample

A company records daily sales:

₹10,000, ₹15,000, ₹20,000, ₹25,000, ₹30,000

Range:

$$ 30,000 - 10,000 = ₹20,000 $$

The company's daily sales vary across a range of ₹20,000

AI/ML Example:                                               #AIMLExample

Suppose a dataset contains customer ages:

18, 22, 30, 45, 60

Range:

 60-18 = 42 

This tells us the overall spread of the age feature
```

```python
data = [10, 20, 30, 40, 50]                         #PythonImplementation

range_value = max(data) - min(data)

print("Range:", range_value)
```

### Variance

$$
\sigma^2 = \frac{\sum_{i=1}^{N}(x_i-\mu)^2}{N}
$$

Where:

\(x_i\) = individual value
\(\mu\) = population mean
\(N\) = number of values
\(\sigma^2\) = population variance

```python
Variance measures how far the data values are spread from the mean

It does this by calculating the average squared deviation from the mean

Variance tells us how much the values differ from the average

Numerical Example:                          #NumericalExample

Data:
2, 4, 6

Step 1: Find the mean

        2 + 4 + 6
Mean =  ---------- = 4
           3

Step 2: Find deviations

2 - 4 = -2
4 - 4 =  0
6 - 4 =  2

Step 3: Square the deviations

(-2)² = 4
(0)²  = 0
(2)²  = 4

```

Step 4: Calculate variance
$$ Variance = \frac{4+0+4}{3} $$ $$ = \frac{8}{3} $$ $$ = 2.67 $$

Variance ≈ 2.67

```python
Business Example:                                        #BusinessExample

A company compares the variability of daily sales

If:

Variance = 100

sales are relatively consistent.

If:

Variance = 10,000

sales fluctuate much more

AI/ML Example:                                            #AIMLExample

Variance is extremely important in machine learning.

For example, consider a feature:

Age

Variance tells us how much the ages differ from their average.

Variance is also part of the bias-variance tradeoff:

High variance → model may overfit
High bias → model may underfit
```

```python
                                                           #Python Implementation:
#For population variance:

import statistics

data = [2, 4, 6]

variance = statistics.pvariance(data)

#For Sample Variance

print("Variance:", variance)

variance = statistics.variance(data)

print("Sample Variance:", variance)
```

### Standard Deviation

$$
\sigma = \sqrt{\frac{\sum_{i=1}^{N}(x_i-\mu)^2}{N}}
$$

```python
Standard deviation measures the typical spread of data around the mean

It is simply the square root of variance

Numerical Example

Using:

2, 4, 6

We already calculated:

Variance=2.67
```

Therefore:

$$ SD = \sqrt{2.67} $$
SD≈1.63

Business Example:                                         #BusinessExample

Suppose two stores have the same average daily sales:

Store A → Mean = ₹50,000
Store B → Mean = ₹50,000

But:

Store A → SD = ₹2,000
Store B → SD = ₹15,000

Store A has much more consistent sales

Store B has much more variable sales

AI/ML Example:                                            #AIMLExample

Standard deviation is commonly used during feature scaling/standardization.

The standardization formula is:

$$ z = \frac{x-\mu}{\sigma} $$

This converts a feature into a standardized scale

### Quartiles

Quartiles divide ordered data into four parts.

There are three important quartiles:

Quartile	 Meaning
 Q1	         25th percentile
 Q2	         50th percentile / Median
 Q3	         75th percentile

So:

0% -------- 25% -------- 50% -------- 75% -------- 100%
             Q1           Q2           Q3

```python
Numerical Example:

Data:

10, 20, 30, 40, 50, 60, 70

Median:

40

Therefore:

  Q2 = 40 

Lower half:
10, 20, 30

So:

  Q1 = 20 

Upper half:

50, 60, 70

So:

 Q3 = 60 

Therefore:

Q1 = 20

Business Example:                                              #BusinessExample

A company studies employee salaries.

Quartiles can tell management:

Q1 → salary level below which about 25% of employees fall
Q2 → median salary
Q3 → salary level below which about 75% of employees fall

AI/ML Example:                                                 #AIMLExample

Quartiles are useful during Exploratory Data Analysis (EDA).

They help us understand:

Distribution
Spread
Skewness
Potential outliers

They are also used to construct box plots
```

```python
!pip install numpy
```

```python
import numpy as np                      #PythonImplementation

data = [10, 20, 30, 40, 50, 60, 70]

Q1 = np.percentile(data, 25)
Q2 = np.percentile(data, 50)
Q3 = np.percentile(data, 75)

print("Q1:", Q1)
print("Q2:", Q2)
print("Q3:", Q3)
```

```python
import → tells Python to load a library

numpy is a Python library used for numerical calculations

np is simply a short name (alias) for NumPy
```

### Percentiles

$$
P_k = \text{Value below which approximately } k\% \text{ of observations fall}
$$

```python
A percentile tells us the value below which a certain percentage of observations fall

For example:

90th percentile means approximately 90% of observations are at or below that value

Common Percentiles:

     Percentile              Meaning
         P25	               Q1
         P50	               Median
         P75	               Q3
         P90                   90% of data is at/below this value
         P95	               95% of data is at/below this value
         P99	               99% of data is at/below this value

Numerical Example:                              #NumericalExample

Suppose students' scores are:

10, 20, 30, 40, 50, 60, 70, 80, 90

The 50th percentile is approximately:

50
because it is the median

Business Example:                               #BusinessExample

An e-commerce company measures delivery time.

Suppose:

90th percentile = 5 days

This means approximately 90% of deliveries take 5 days or less under the percentile calculation used

This is often more informative than simply looking at average delivery time

AI/ML Example:                                           #AIMLExample

Percentiles are useful for outlier detection and feature analysis

For example:

P99 of transaction amount = ₹50,000

Transactions above this level may deserve additional investigation, depending on the business context
```

```python
import numpy as np                                    #Python Implementation

scores = [10, 20, 30, 40, 50, 60, 70, 80, 90]

p90 = np.percentile(scores, 90)

print("90th Percentile:", p90)
```

### Interquartile Range (IQR)

$$
IQR = Q_3 - Q_1
$$

```python
Interquartile Range (IQR) measures the spread of the middle 50% of the data

Numerical Example

Suppose:

 Q1 = 20 

and:

 Q3 = 60 

Then:

 IQR = 60-20 

 IQR = 40

Business Example

Suppose customer spending has:

Q1 = ₹1,000
Q3 = ₹5,000

Then:

 IQR = 5000-1000 = ₹4,000 

The middle 50% of customers have spending spread across ₹4,000

AI/ML Example:

IQR is commonly used for outlier detection.

The common rule is:

 Lower  Bound = Q1 - 1.5(IQR) 
 Upper Bound =  Q3 + 1.5(IQR) 

Values outside these bounds are often treated as potential outliers
```

```python
import numpy as np                                     #PythonImplementation

data = [10, 20, 30, 40, 50, 60, 70]

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

### Coefficient of Variation

$$
CV = \frac{Standard\ Deviation}{Mean} \times 100
$$

```python
Coefficient of Variation (CV) measures the standard deviation relative to the mean

It is especially useful for comparing variability between datasets with different means or different scales

Numerical Example                                #NumericalExample

Suppose:

Mean = 50
Standard Deviation = 5
       5 
CV =  ---- * 100
       50

CV=10%

Business Example:                    #BusinessExample

Suppose two products have:

Product A

Mean sales = ₹10,000
SD = ₹1,000

CVA = 10000/1000 ×100  =10%

Product B

Mean sales = ₹50,000
SD = ₹10,000

   CVB = /5000010000×100 =20%
       
 Although Product B has higher absolute sales, its sales are more variable relative to its average

 AI/ML Example                       AIMLExample

Suppose two features have:

Feature	    Mean	SD
Feature A	100	   10
Feature B	20	    5      

     Feature A
         
     CV = 100/10×100 =10%

    Feature B

    CV = 20/5×100 =25%            

 Feature B has greater relative variability

CV can therefore help compare variability across features measured on different scales, although it is not appropriate when the mean is zero or very close to zero                 
```

```python
import statistics                   #PythonImplementation

data = [45, 48, 50, 52, 55]

mean = statistics.mean(data)
sd = statistics.pstdev(data)

cv = (sd / mean) * 100

print("Mean:", mean)
print("Standard Deviation:", sd)
print("CV:", cv, "%")
```
