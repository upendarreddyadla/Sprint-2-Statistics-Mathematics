## T-Test

```python
A T-Test is used to determine whether there is a statistically significant difference between the means of groups

It is commonly used when the sample size is relatively small and the population standard deviation is unknown

Types of T-Test:

1.One-sample T-test → Compare one sample mean with a known value

2.Independent two-sample T-test → Compare means of two independent groups

3.Paired T-test → Compare measurements from the same subjects before and after something

Suppose:

Group A average salary = ₹40,000
Group B average salary = ₹45,000

The T-test tells us whether the ₹5,000 difference is statistically significant

Assumptions:

Data is quantitative
Observations are independent for an independent T-test
Data is approximately normally distributed, especially for small samples
For the standard independent T-test, variances are assumed equal
For unequal variances, Welch's T-test can be used

When to use:

Use a T-test when:

Comparing means
There are one or two groups
Population standard deviation is unknown
Data is approximately normally distributed

When NOT to use:

Don't use a T-test when:

Comparing more than two independent groups → use ANOVA.
Data is strongly non-normal with small samples → consider a non-parametric test.
Data is categorical → use Chi-Square.
Groups are paired but you use an independent test

Business Example:

A company wants to know whether a new training program increased employee productivity.

Before training:
[60, 62, 65, 61, 59]

After training:
[70, 72, 68, 75, 71]

We can use a paired T-test
```

```python
from scipy.stats import ttest_rel                          #PythonExample

before = [60, 62, 65, 61, 59]
after = [70, 72, 68, 75, 71]

t_stat, p_value = ttest_rel(before, after)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
if p_value < 0.05:                               #Interpretation
    print("Reject H0: Training had a significant effect.")
else:
    print("Fail to reject H0: No significant effect.")
```

## Z-Test

A Z-Test determines whether a sample mean is significantly different from a population mean or whether two proportions/means differ significantly

It is generally used when the population standard deviation is known or when the sample size is sufficiently large under appropriate assumptions

Formula:

$$ Z = \frac{\bar{x}-\mu}{\sigma/\sqrt{n}} $$

Where:

x̄ = sample mean
μ = population mean
σ = population standard deviation
n = sample size

### Assumptions:

Data is quantitative for a mean Z-test.
Observations are independent.
Population standard deviation is known for the classical one-sample mean Z-test.
Population is normally distributed, or sample size is sufficiently large for the normal approximation.

### When to use

Use a Z-test when:

Population standard deviation is known.
Sample size is large.
Testing a population mean or proportion

### When NOT to use

Don't use a Z-test when:

Population standard deviation is unknown and sample size is small
Data is heavily non-normal with a small sample
You need to compare more than two means → use ANOVA

### Business Example:

A company claims:

Average delivery time = 30 minutes.

You collect data from 100 deliveries and want to test this claim.

Population mean = 30
Sample mean = 32
Population SD = 8
n = 100

```python
from statsmodels.stats.weightstats import ztest              #PythonExample

sample = [
    32, 31, 30, 33, 29,
    34, 31, 32, 30, 33
]

z_stat, p_value = ztest(sample, value=30)

print("Z-statistic:", z_stat)
print("P-value:", p_value)
```

```python
if p_value < 0.05:                    #Interpretation
    print("Reject H0")
else:
    print("Fail to reject H0")
```

## Chi-Square Test

The Chi-Square test is mainly used for categorical data

It checks whether:

Two categorical variables are related (Chi-Square Test of Independence), or
Observed category counts match expected counts (Goodness-of-Fit)

Formula:

$$ \chi^2 = \sum \frac{(O-E)^2}{E} $$

Where:

O = Observed frequency
E = Expected frequency

### Assumptions:

Data is categorical/count data
Observations are independent
Expected frequencies should generally be sufficiently large
Categories should be mutually exclusive

### When to use

Use Chi-Square when:

Both variables are categorical
You want to test association between categories
You want to compare observed and expected frequencies

### When NOT to use

Don't use it when:

Data is continuous without categorization
Expected counts are very small
Observations are not independent
You are comparing numerical means

### Business Example

A company wants to know whether customer gender is associated with product preference.

	     Product A	 Product B
         
Male	      40	   20

Female	      25	   35

Both variables are categorical

```python
from scipy.stats import chi2_contingency                #pythonExample

data = [
    [40, 20],
    [25, 35]
]

chi2, p_value, dof, expected = chi2_contingency(data)

print("Chi-square:", chi2)
print("P-value:", p_value)
print("Degrees of freedom:", dof)
print("Expected frequencies:")
print(expected)
```

```python
if p_value < 0.05:                                          #Interpretation
    print("Reject H0: Variables are associated")
else:
    print("Fail to reject H0: No significant association")
```

## ANOVA

ANOVA — Analysis of Variance is used to compare the means of three or more groups

For example:

Marketing Campaign A → Sales
Marketing Campaign B → Sales
Marketing Campaign C → Sales

Main idea:

ANOVA compares:

Between-group variation
          /
         /
Within-group variation

The F-statistic is:


       Within-group variation
F =    -----------------------
       Between-group variation


### Assumptions:

Dependent variable is numerical
Groups are independent
Observations are independent
Data within groups is approximately normally distributed
Variances are approximately equal across groups for classical one-way ANOVA

### When to use

Use ANOVA when:

Comparing 3 or more means
Groups are independent
Numerical dependent variable
Assumptions are reasonably satisfied

### When NOT to use

Don't use standard one-way ANOVA when:

There are only two groups → T-test is usually simpler.
Data is severely non-normal and assumptions are badly violated.
Groups are dependent/repeated measures → use repeated-measures ANOVA or another appropriate method.
Dependent variable is categorical.

### Business Example

A company tests three advertising campaigns.

Campaign A → Sales
Campaign B → Sales
Campaign C → Sales

Question:

Is there a significant difference in average sales among the three campaigns

### Interpretation

If:

p-value < 0.05

then:

Reject H0

Conclusion: At least one campaign has a different mean sales level

ANOVA does not tell you exactly which groups differ

After significant ANOVA, you can perform a post-hoc test, such as Tukey's HSD

```python
from scipy.stats import f_oneway                          #PythonExample

campaign_A = [20, 22, 21, 19, 23]
campaign_B = [25, 27, 26, 28, 24]
campaign_C = [30, 32, 31, 29, 33]

f_stat, p_value = f_oneway(
    campaign_A,
    campaign_B,
    campaign_C
)

print("F-statistic:", f_stat)
print("P-value:", p_value)
```

## Mann-Whitney U Test


The Mann-Whitney U test is a non-parametric test used to compare two independent groups

It is an alternative to the independent two-sample T-test when the assumptions for the T-test are not appropriate

Instead of directly comparing means, it works with ranks.

### Assumptions:

Two independent groups
Observations are independent
Outcome is at least ordinal
Data can be ranked
For the common interpretation as a distribution/location comparison, the groups should have reasonably similar distribution shapes

### When to use

Use Mann-Whitney U when:

Comparing two independent groups
Data is ordinal or continuous
Data is not normally distributed
Outliers make a T-test inappropriate
Sample size is small and normality is questionable

### When NOT to use

Don't use it when:

Groups are paired → use Wilcoxon signed-rank test
You have more than two independent groups → consider Kruskal-Wallis
Data is categorical with counts → use Chi-Square
You specifically need to compare means

Important: Mann-Whitney U is not simply a universal "test of medians." Its interpretation depends on the distributions of the two groups

### Business Example:

A company compares customer satisfaction scores between two branches.

Branch A:
[7, 8, 6, 9, 7]

Branch B:
[4, 5, 6, 5, 4]

The satisfaction data may not be normally distributed

### Interpretation

If:

p-value = 0.02

then:

0.02 < 0.05

Therefore:

Reject H₀.

There is statistically significant evidence that the two groups differ in their distributions/ranks

```python
from scipy.stats import mannwhitneyu               #PythonExample

branch_A = [7, 8, 6, 9, 7]
branch_B = [4, 5, 6, 5, 4]

u_stat, p_value = mannwhitneyu(
    branch_A,
    branch_B,
    alternative="two-sided"
)

print("U-statistic:", u_stat)
print("P-value:", p_value)
```

## Shapiro-Wilk Test

The Shapiro-Wilk test checks whether a sample is consistent with a normal distribution

It is commonly used as a normality check before applying statistical tests that assume normality

Hypotheses

H₀

Data follows a normal distribution

H₁

Data does not follow a normal distribution

### Assumptions:

Observations should be independent
Data should be quantitative/continuous for typical use
The test is designed for assessing normality of a sample
Very large samples can make even tiny departures from normality statistically significant

### When to use

Use Shapiro-Wilk when:

You need to check normality
You are deciding whether a parametric test's normality assumption is reasonable
You have a relatively small/moderate sample

### When NOT to use

Don't rely on Shapiro-Wilk alone when:

The sample is extremely large
You are checking categorical data
You need to determine equal variance — use an appropriate variance test instead
You are treating a non-significant p-value as proof that data is perfectly normal

It's often useful to combine it with a Q-Q plot and domain knowledge.

### Business Example

Suppose a company has employee salary data.

Before using a T-test or ANOVA, we want to check whether the data is approximately normal.

### Interpretation

If:

p-value = 0.20

then:

0.20 > 0.05

We fail to reject H₀.

Conclusion: There is not sufficient statistical evidence to say that the sample departs from normality.

If:

p-value = 0.01

then:

0.01 < 0.05

We reject H₀

Conclusion: There is evidence that the data is not normally distributed

```python
from scipy.stats import shapiro               #PythonExample

salary = [
    30000, 32000, 31000, 35000,
    33000, 34000, 36000, 31500
]

stat, p_value = shapiro(salary)

print("Shapiro-Wilk statistic:", stat)
print("P-value:", p_value)
```
