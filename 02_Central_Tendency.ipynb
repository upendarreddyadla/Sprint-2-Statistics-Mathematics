### Mean

```python
Mean is the average value of a set of numbers
Mean = Sum of all values ÷ Number of values

Formula:    
                           
	      Number of values
    Mean= ---------------
          Sum of all values

Easy way to remember

Mean = Average

Example:- Numercal                                                #Numerical Example

Suppose the marks are:

10, 20, 30, 40, 50

Step 1: Add all values

10 + 20 + 30 + 40 + 50 = 150

Step 2: Count the values

There are 5 values

Step 3: Divide

Mean = 150 ÷ 5 = 30

So, the Mean = 30

Business Example:                                                   #Business Example

A company has monthly sales:

₹10,000, ₹20,000, ₹30,000, ₹40,000, ₹50,000

The mean sales are:

₹30,000

This tells the company its average monthly sales


AI/ML Example:                                                   #AIML Example                             

Suppose a dataset contains customer ages:

20, 25, 30, 35, 40

Mean age = 30

The mean can be used for data analysis and preprocessing

Interpretation:                                                   #Interpretation

A mean of 30 means:

The dataset has an average value of 30.

Important: Mean can be strongly affected by extreme values (outliers)    
```

```python
data = [10, 20, 30, 40, 50]                    #Python Implementation of Mean

mean = sum(data) / len(data)

print("Mean:", mean)
```

### Median

```python
Median is the middle value in a dataset when the values are arranged in ascending or descending order
Median = The middle value of ordered data

Odd → One middle value
Even → Average of two middle values

Example:

Data:

10, 20, 30, 40, 50

There are 5 values

The middle value is 30.

So,The Median is = 30

Example 2: Even number of values

Data:

10, 20, 30, 40

There are 4 values, so there is no single middle value

Take the two middle values:

20 and 30

Then calculate their average:

        
          20 + 30
 Median= ---------
            2

 Median = 25

Steps to find Median:
     
1.Arrange the data in ascending order
2.Count the number of values
3.If the number is odd → choose the middle value
4.If the number is even → take the average of the two middle values
```
